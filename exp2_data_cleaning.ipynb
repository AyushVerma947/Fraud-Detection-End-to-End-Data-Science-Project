```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("data/online_fraud.csv")

print("Original Shape:", df.shape)
```

```python
df = df.sample(n=100000, random_state=42)
print("Sampled Shape:", df.shape)
```

```python
df.isnull().sum()
```

```python
df.loc[df.sample(frac=0.05).index, 'amount'] = np.nan
df.loc[df.sample(frac=0.05).index, 'oldbalanceOrg'] = np.nan

df.isnull().sum()
```

```python
mean_imputer = SimpleImputer(strategy='mean')

df[['amount', 'oldbalanceOrg']] = mean_imputer.fit_transform(
    df[['amount', 'oldbalanceOrg']]
)

df.isnull().sum()
```

```python
df.drop(['nameOrig', 'nameDest'], axis=1, inplace=True)
```

```python
df = pd.get_dummies(df, columns=['type'], drop_first=True)

df.head()
```

```python
df.dtypes
```

```python
scaler = StandardScaler()

numerical_cols = ['step', 'amount', 'oldbalanceOrg',
                  'newbalanceOrig', 'oldbalanceDest',
                  'newbalanceDest']

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

df.head()
```

```python
print("Final Shape:", df.shape)
print("Any Missing Values?", df.isnull().sum().sum())
```
